```python
%%manim -v WARNING -qh --format=mp4 HypergraphAnimation

from manim import *

from PIL import Image
import numpy as np
from pathlib import Path

frames_dir = Path("frames")
frames_dir.mkdir(exist_ok=True)


def save_frame(scene, name):
    img = Image.fromarray(scene.renderer.get_frame())
    img.save(frames_dir / f"{name}.png")
    
class HypergraphAnimation(Scene):
    def construct(self):
        self.camera.background_opacity = 0
        
        # 1. Define Positions
        pos = {
            'a': np.array([-3.0, 1.5, 0]),
            'b': np.array([-1.0, 1.7, 0]),
            'c': np.array([-2.0, 0.0, 0]),
            'd': np.array([0.3, 1.8, 0]),
            'e': np.array([1.6, 1.5, 0]),
            'f': np.array([0.0, -0.1, 0]),
            'g': np.array([-0.2, -2.0, 0]),
        }

        # 3. Create Default Vertices
        nodes = {}
        for name, point in pos.items():
            circle = Circle(radius=0.3, color=WHITE, stroke_color=BLACK, stroke_width=2, fill_color=WHITE, fill_opacity=1)
            label = MathTex(name, color=BLACK).scale(0.8)
            node = VGroup(circle, label).move_to(point)
            nodes[name] = node

        all_nodes = VGroup(*nodes.values())

        # 2. Define Hyperedges
        hedge1 = Polygon(
            pos['a'] + [-0.65, 0.40, 0],
            pos['b'] + [ 0.65, 0.50, 0],
            pos['c'] + [ 0.00,-0.7, 0],
            color="#1b4d2e",
            fill_color="#408055",
            fill_opacity=0.22,
            stroke_width=2,
        ).round_corners(0.35)
        
        hedge2 = Polygon(
            pos['b'] + [-0.65, 0.20, 0],
            pos['d'] + [ 0.45, 0.55, 0],
            pos['f'] + [ 0.25, -1, 0],
            color="#1b4d2e",
            fill_color="#408055",
            fill_opacity=0.22,
            stroke_width=2,
        ).round_corners(0.35)
        
        hedge3 = Polygon(
            pos['c'] + [-0.8, 0.45, 0],
            pos['f'] + [ 0.55, 0.35, 0],
            pos['g'] + [ 0.25,-0.85, 0],
            color="#1b4d2e",
            fill_color="#408055",
            fill_opacity=0.22,
            stroke_width=2,
        ).round_corners(0.35)

        hyperedges = VGroup(hedge1, hedge2, hedge3)
        graph = VGroup(hedge1, hedge2, hedge3, *all_nodes)

        graph.scale_to_fit_width(config.frame_width * 0.6)
        
        # 4. Animate Diagram Construction
        self.play(FadeIn(all_nodes, lag_ratio=0.1), run_time=1)
        save_frame(self, "frame0")
        
        self.play(Create(hedge1), run_time=0.8)
        save_frame(self, "frame1")
        
        self.play(Create(hedge2), run_time=0.8)
        save_frame(self, "frame2")
        
        self.play(Create(hedge3), run_time=0.8)
        save_frame(self, "frame3")
        
        self.wait(0.5)

        # Helper for green nodes
        def make_sampled_node(name):
            circle = Circle(radius=0.35, color=WHITE, fill_color="#0B3D20", fill_opacity=1)
            label = MathTex(name, color=WHITE).scale(0.8)
            return VGroup(circle, label).move_to(pos[name])

        # 5. Animate Sampling Sequence (FIXED PULSE ANIMATION)
        sampled_sequence = ['a', 'b', 'd', 'e', 'g']

        for i, name in enumerate(sampled_sequence, start=4):
            circle = nodes[name][0]
            label = nodes[name][1]

            self.play(
                circle.animate.set_color("#1B5E20"),
                label.animate.set_color(WHITE),
            )
            
            self.play(
                nodes[name].animate.scale(1.1),
                run_time=0.3,
            )
            save_frame(self, f"frame{i}")
            self.wait(0.2)

        self.wait(1.5)
```
